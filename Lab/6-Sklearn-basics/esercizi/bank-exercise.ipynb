```python
# Execute once
df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})
df
```

## Es. 3

```python
oh_enc = sk.preprocessing.OneHotEncoder()
```

```python
oh_enc.fit(df['Geography'].to_numpy().reshape(-1,1))
```

```python
enc_data = oh_enc.transform(df['Geography'].to_numpy().reshape(-1,1)).toarray()
enc_data
```

```python
cols = oh_enc.get_feature_names_out()
cols
```

```python
df_geography_enc = pd.DataFrame(enc_data, columns=cols)
```

```python
df = df.drop(columns=['Geography'], axis=1)
df
```

```python
df = pd.concat([df, df_geography_enc], axis=1)
df
```

## Es. 4

These three columns must be removed:
- *RowNumber* and *CustomerId* must be removed to prevent overfitting;
- *Surname* both for overfitting and because it is something that really should not contribute to the prediction.

```python
df = df.drop(columns=['RowNumber', 'CustomerId', 'Surname'])
df
```

```python
X = df.drop(columns=['Exited'])
X
```

```python
y = df[['Exited']]
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.75, random_state=0)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

## Es. 5

```python
from sklearn.tree import DecisionTreeClassifier
dtc = DecisionTreeClassifier()
```

```python
dtc.fit(X_train, y_train)
```

```python
y_pred = dtc.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

acc = accuracy_score(y_test, y_pred)
print(f'Accuracy: {acc*100}%')
```
