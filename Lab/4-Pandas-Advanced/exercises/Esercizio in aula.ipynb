2. Crea una nuova colonna chiamata **Livello Esperienza**, utilizzando la colonna **Anni di esperienza**. Se l'esperienza è minore di 5 anni, mappa il valore a "Junior", tra 5 e 10 a "Mid" e più di 10 a "Senior".

```python
dipendenti['Livello Esperienza'] = dipendenti['Anni di esperienza']
dipendenti
```

```python
def find_level(anni):
    if anni < 5:
        return 'Junior'
    elif anni <= 10:
        return 'Mid'
    else:
        return 'Senior'
```

```python
dipendenti['Livello Esperienza'] = dipendenti['Anni di esperienza'].map(find_level)
dipendenti
```

```python
dipendenti['Livello Esperienza cut'] = pd.cut(dipendenti['Anni di esperienza'].values, [0, 5, 11, 100], right=False, labels=['Junior', 'Mid', 'Senior'])
dipendenti
```

3. Sostituisci i nomi dei dipartimenti con le seguenti abbreviazioni: Vendite -> 'VEN', Information Technology -> 'IT', Human Resources -> 'HR'. Quanti dipendenti ci sono per ogni dipartimento?

```python
abbr = {'Vendite': 'VEN', 'Information Technology': 'IT', 'Human Resources': 'HR'}
dipendenti['Dipartimento'] = dipendenti['Dipartimento'].map(abbr)
dipendenti
```

4. Crea un'altra colonna con i valori di età discretizzati in 5 gruppi, a quale gruppo di età appartiene la maggior parte dei dipendenti?

```python
dipendenti['Età Discr'] = pd.cut(dipendenti['Età'].values, 5)
dipendenti
```

```python
dipendenti['Età Discr'] = pd.cut(dipendenti['Età'].values, 5).codes
dipendenti
```

5. Sostituisci i valori mancanti della colonna età con il valore medio di età per i dipendenti dello stesso settore.

```python
medie = {}
#medie.append(dipendenti['Età'][])
medie['VEN'] = dipendenti['Età'].where(dipendenti['Dipartimento'] == 'VEN').mean()
medie['IT'] = dipendenti['Età'].where(dipendenti['Dipartimento'] == 'IT').mean()
medie['HR'] = dipendenti['Età'].where(dipendenti['Dipartimento'] == 'HR').mean()
medie
```

```python
dipendenti['Età'].fillna(medie['VEN'], inplace=True)
dipendenti['Età'].fillna(medie['IT'], inplace=True)
dipendenti['Età'].fillna(medie['HR'], inplace=True)
dipendenti
```
