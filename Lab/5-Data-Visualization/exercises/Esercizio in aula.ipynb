## 2. Aggiungere al dataset:

- una nuova colonna "temp_diff" con la differenza in valore assoluto tra i valori di temperatura alle 9 del mattino e le 3 del pomeriggio

- una nuova colonna "hum_diff" con la differenza in valore assoluto tra i valori di umidità alle 9 del mattino e le 3 del pomeriggio

```python
df['temp_diff'] = np.abs(df.Temp9am - df.Temp3pm)
df['hum_diff'] = np.abs(df.Humidity9am - df.Humidity3pm)
df
```

## 3. Selezionare soltanto le righe in cui "temp_diff" è maggiore di 10 e "hum_diff" è maggiore di 20. Qual è la location in cui si verifica più spesso questa elevata escursione termica (ovvero la location con il maggior numero di righe nella selezione)?

```python
df[(df.temp_diff>10) & (df.hum_diff > 20)].groupby('Location')['Location'].count().idxmax()
```

```python
filtered_df = df[(df['temp_diff'] > 10) & (df['hum_diff'] > 20)]
```

## 5. Trasformare i valori della colonna "RainToday" convertendo "yes" in 1 e "no" in 0. Utilizzando il metodo groupby per calcolare per ogni location il numero di giorni di pioggia.

```python
mapping = {'Yes': 1, 'No': 0}
# Execute only once, otherwise RainToday will be filled with missing values, because at the second time Yes and No won't be found
df['RainToday'] = df['RainToday'].map(mapping)
df
```

```python
df[['Location', 'RainToday']].groupby('Location').sum()
```

```python
#df[['RainToday']].groupby('Location').sum()
# ATTENTION: this does not work, as you're trying to group on column 'Location', but you've already filtered it out.
# It works if you first group and then filter by column:
df.groupby('Location')['RainToday'].sum()
```

## 6. Creare una tabella pivot in cui per ogni location (sulle righe) e per ogni mese (sulle colonne) sia calcolata la temperatura massima media.

```python
df.pivot_table(index=['Location'], columns=['Month'], values=['MaxTemp'], aggfunc='mean')
```

## 7. Visualizzare in un unico grafico la distribuzione dei valori di temperatura massima nel mese di maggio (5) e dicembre (12).

```python
max_temps_may = df[df['Month'] == 5]['MaxTemp']#.values
max_temps_dec = df[df['Month'] == 12]['MaxTemp']#.values
max_temps_may, max_temps_dec
```

```python
fig, ax = plt.subplots()
ax.set_ylabel('Max Temperature (°C)')
splot = ax.boxplot([max_temps_may, max_temps_dec],
                    widths= 0.3,
                    patch_artist=True,  # fill with color
                    tick_labels=['May', 'December'] # will be used to label x-ticks
                    )  
                    
```

Alternativa con istogramma

```python
plt.hist(max_temps_may, alpha=0.5, label='Maggio')
plt.hist(max_temps_dec, alpha=0.5, label='Dicembre')
plt.legend(loc='upper right')
plt.xlabel('Temperatura massima')
plt.ylabel('Frequenza')
plt.show()
```

## 8. Visualizzare in uno scatterplot la relazione tra i valori di pressione e i valori di umidità (fare 2 plot distinti nella stessa figura, uno per i valori 9am e un altro per i valori 3pm). Colorare i punti di due colori diversi in base ai valori di RainToday.

```python
fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(10,5))
s = axs[0].scatter(df.Pressure9am, df.Humidity9am, c=df.RainToday)
axs[0].set_xlabel('Pressure')
axs[0].set_ylabel('Humidity')
axs[0].set_title('9 AM')
axs[0].legend(handles=s.legend_elements()[0], labels=s.legend_elements()[1], loc='best')

axs[1].scatter(df.Pressure3pm, df.Humidity3pm, c=df.RainToday)
axs[1].set_xlabel('Pressure')
axs[1].set_ylabel('Humidity')
axs[1].set_title('3 PM')
axs[1].legend(handles=s.legend_elements()[0], labels=s.legend_elements()[1], loc='best')
```

Ora rimuovo gli outlier che ci sono nel secondo subplot, per vedere meglio i dati. Il resto è uguale.

```python
df['Pressure3pm'] = np.where(df['Pressure3pm'] < 0, np.nan, df['Pressure3pm'])
df.dropna(inplace=True)

fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(10,5))
s = axs[0].scatter(df.Pressure9am, df.Humidity9am, c=df.RainToday)
axs[0].set_xlabel('Pressure')
axs[0].set_ylabel('Humidity')
axs[0].set_title('9 AM')
axs[0].legend(handles=s.legend_elements()[0], labels=s.legend_elements()[1], loc='best')

axs[1].scatter(df.Pressure3pm, df.Humidity3pm, c=df.RainToday)
axs[1].set_xlabel('Pressure')
axs[1].set_ylabel('Humidity')
axs[1].set_title('3 PM')
axs[1].legend(handles=s.legend_elements()[0], labels=s.legend_elements()[1], loc='best')
```
